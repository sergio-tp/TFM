En este cuaderno, para poder trabajar con el archivo de tasación que se encuentra en formato .XLS, transformaremos los datos a .csv para facilitar su lectura, y poder trabajar con ellos con spark fácilmente.

```python
import pandas as pd

archivo_excel = "../datasets/dataset_appraisals.XLS"

hojas = pd.read_excel(
    archivo_excel,
    sheet_name=None
)

for i, (_, df) in enumerate(hojas.items(), start=1):
    df.to_csv(
        f"../datasets/dataset_appraisals{i}.csv",
        index=False
    )
```
